### 1. Carga del Dataset y Verificación Estructural

```python
import os
import pandas as pd
import numpy as np

# Cargar archivo en Google Colab
file_name = 'dataset_pm25.csv'
if not os.path.exists(file_name):
    print('Por favor sube el archivo dataset_pm25.csv:')
    from google.colab import files
    uploaded = files.upload()

df = pd.read_csv(file_name)
print('=' * 60)
print(f'✓ Dataset cargado exitosamente.')
print(f'✓ Total filas: {df.shape[0]:,}')
print(f'✓ Total columnas: {df.shape[1]}')
print(f'✓ Total celdas analizadas: {df.size:,}')
print('=' * 60)
df.head()
```

### 2. Control de Calidad: Verificación de Valores Nulos (NaN)

```python
null_audit = pd.DataFrame({
    'Tipo de Dato': df.dtypes.astype(str),
    'Registros Válidos': df.count(),
    'Valores Nulos (NaN)': df.isnull().sum(),
    '% Nulos': (df.isnull().sum() / len(df) * 100).round(2)
})

null_audit.style.set_caption('Control de Calidad: Detección de Valores Nulos por Variable') \
    .set_table_styles([{
        'selector': 'th',
        'props': [('background-color', '#1F4E79'), ('color', 'white'), ('text-align', 'center'), ('font-weight', 'bold')]
    }]) \
    .format({'Registros Válidos': '{:,}', 'Valores Nulos (NaN)': '{:,}', '% Nulos': '{:.2f}%'}) \
    .set_properties(**{'text-align': 'center', 'border': '1px solid #D9D9D9'})
```

### 3. Matriz Paramétrica de Estadísticas Descriptivas (Equivalente Excel)

```python
metrics_dict = {
    'Registros válidos (N)': df.count(),
    'Valores faltantes (NaN)': df.isnull().sum(),
    'Media aritmética (μ)': df.mean(),
    'Desviación estándar (s)': df.std(),
    'Mínimo (Min)': df.min(),
    'Primer cuartil (Q25)': df.quantile(0.25),
    'Mediana (Q50)': df.median(),
    'Tercer cuartil (Q75)': df.quantile(0.75),
    'Máximo (Max)': df.max(),
    'Rango Intercuartílico (IQR)': df.quantile(0.75) - df.quantile(0.25),
    'Rango Total (Max - Min)': df.max() - df.min()
}

df_stats = pd.DataFrame(metrics_dict).T

styled_stats = df_stats.style \
    .set_caption('MATRIZ DE AUDITORÍA ESTADÍSTICA Y CONTROL DE CALIDAD - BEIJING PM2.5 DATASET (41.730 REGISTROS)') \
    .set_table_styles([
        {'selector': 'caption', 'props': [('font-size', '14px'), ('font-weight', 'bold'), ('color', '#1F4E79'), ('padding-bottom', '8px')]},
        {'selector': 'th', 'props': [('background-color', '#1F4E79'), ('color', 'white'), ('font-weight', 'bold'), ('text-align', 'center'), ('padding', '6px 10px')]},
        {'selector': 'tbody tr:nth-child(even)', 'props': [('background-color', '#F9FAFB')]},
        {'selector': 'tbody tr:nth-child(odd)', 'props': [('background-color', '#FFFFFF')]},
        {'selector': 'td', 'props': [('border', '1px solid #D9D9D9'), ('padding', '6px 10px'), ('text-align', 'right')]}
    ]) \
    .format(lambda v: f'{v:,.0f}' if abs(v) >= 1000 and abs(v - round(v)) < 1e-4 else f'{v:,.2f}')

styled_stats
```
